```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("RetailIQ Dashboard Preparation started!")
```

```python
# ===== LOAD PROCESSED DATA =====

sales_df = pd.read_csv(
    "../data/processed/clean_sales_data.csv"
)

rfm = pd.read_csv(
    "../data/processed/rfm_customer_segments.csv"
)

product_performance = pd.read_csv(
    "../data/processed/product_performance.csv"
)

country_performance = pd.read_csv(
    "../data/processed/country_performance.csv"
)

forecast = pd.read_csv(
    "../data/processed/revenue_forecast.csv"
)

sales_df["InvoiceDate"] = pd.to_datetime(sales_df["InvoiceDate"])
forecast["Month"] = pd.to_datetime(forecast["Month"])

print("All dashboard datasets loaded successfully!")
```

```python
# ===== DASHBOARD KPIs =====

total_revenue = sales_df["Revenue"].sum()
total_orders = sales_df["InvoiceNo"].nunique()
total_customers = sales_df["CustomerID"].nunique()
total_products = sales_df["StockCode"].nunique()

average_order_value = total_revenue / total_orders

dashboard_kpis = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Orders",
        "Total Customers",
        "Total Products",
        "Average Order Value"
    ],
    "Value": [
        total_revenue,
        total_orders,
        total_customers,
        total_products,
        average_order_value
    ]
})

display(dashboard_kpis)
```

```python
# ===== MONTHLY SALES FOR DASHBOARD =====

monthly_sales = (
    sales_df
    .set_index("InvoiceDate")
    .resample("ME")["Revenue"]
    .sum()
    .reset_index()
)

monthly_sales.columns = ["Month", "Revenue"]

display(monthly_sales)
```

```python
# ===== SAVE DASHBOARD DATA =====

dashboard_kpis.to_csv(
    "../data/processed/dashboard_kpis.csv",
    index=False
)

monthly_sales.to_csv(
    "../data/processed/monthly_sales.csv",
    index=False
)

print("Dashboard datasets prepared successfully!")
```
