```python
# ==========================================
# 1. IMPORT LIBRARIES AND LOAD DATASET
# ==========================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load tips.csv
df = pd.read_csv("tips.csv")

# Display first 5 rows
print(df.head())

# Display column names
print("\nColumns:")
print(df.columns)

# Display dataset information
print("\nDataset Information:")
print(df.info())
```

```python
# ==========================================
# 2. HISTOGRAM USING MATPLOTLIB
# ==========================================

import matplotlib.pyplot as plt

plt.hist(df['total_bill'])

plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# 3. HISTOGRAM USING SEABORN
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['total_bill'])

plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# 4. KDE PLOT USING SEABORN
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.kdeplot(
    df['total_bill'],
    fill=True
)

plt.title("KDE Plot of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```

```python
# ==========================================
# 5. KDE PLOT USING PANDAS
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

df['total_bill'].plot(
    kind='kde'
)

plt.title("KDE Plot of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```

```python
# ==========================================
# 6. KDE PLOT WITH MATPLOTLIB
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

# Calculate KDE using seaborn
sns.kdeplot(
    data=df,
    x='total_bill'
)

plt.title("KDE Plot using Matplotlib Visualization")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```

```python
# ==========================================
# 7. COUNT PLOT
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=df,
    x='sex'
)

plt.title("Count Plot of Gender")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python
# ==========================================
# 8. SCATTER PLOT
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip'
)

plt.title("Scatter Plot: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

```python
# ==========================================
# 9. BAR GRAPH
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(
    data=df,
    x='day',
    y='total_bill'
)

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")

plt.show()
```

```python
# ==========================================
# 10. SUBPLOT
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))

# First subplot
plt.subplot(1, 2, 1)

sns.histplot(
    data=df,
    x='total_bill',
    kde=True
)

plt.title("Histogram of Total Bill")

# Second subplot
plt.subplot(1, 2, 2)

sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip'
)

plt.title("Total Bill vs Tip")

plt.tight_layout()

plt.show()
```

```python
# ==========================================
# 11. HISTOGRAM WITH RWIDTH AND ALPHA
# ==========================================

import matplotlib.pyplot as plt

plt.hist(
    df['total_bill'],
    bins=10,
    rwidth=0.8,
    alpha=0.7
)

plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# 12. COUNT PLOT
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=df,
    x='day'
)

plt.title("Number of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Count")

plt.show()
```

```python
# ==========================================
# 13. COUNT PLOT WITH PALETTE
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=df,
    x='day',
    hue='day',
    palette='Set2',
    legend=False
)

plt.title("Count Plot of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Count")

plt.show()
```

```python
# ==========================================
# 14. BOX PLOT
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(
    data=df,
    x='day',
    y='total_bill'
)

plt.title("Box Plot of Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")

plt.show()
```

```python
# ==========================================
# 15. HEATMAP
# ==========================================

import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
numeric_df = df.select_dtypes(
    include='number'
)

# Calculate correlation
correlation = numeric_df.corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# ==========================================
# 16. DASHBOARD
# ==========================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("tips.csv")

# Create dashboard
fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)

# ------------------------------------------
# Dashboard 1: Histogram
# ------------------------------------------

sns.histplot(
    data=df,
    x='total_bill',
    kde=True,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Total Bill Distribution")

# ------------------------------------------
# Dashboard 2: Scatter Plot
# ------------------------------------------

sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    ax=axes[0, 1]
)

axes[0, 1].set_title("Total Bill vs Tip")

# ------------------------------------------
# Dashboard 3: Count Plot
# ------------------------------------------

sns.countplot(
    data=df,
    x='day',
    ax=axes[1, 0]
)

axes[1, 0].set_title("Customers by Day")

# ------------------------------------------
# Dashboard 4: Box Plot
# ------------------------------------------

sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    ax=axes[1, 1]
)

axes[1, 1].set_title("Total Bill by Day")

# Adjust layout
plt.tight_layout()

plt.show()
```

```python
# ==========================================
# 17B. DASHBOARD USING CALLBACK
# ==========================================

import pandas as pd

from dash import Dash, dcc, html, Input, Output
import plotly.express as px

# ------------------------------------------
# Load dataset
# ------------------------------------------

df = pd.read_csv("tips.csv")

# ------------------------------------------
# Create Dash application
# ------------------------------------------

app = Dash(__name__)

# ------------------------------------------
# Dashboard Layout
# ------------------------------------------

app.layout = html.Div([

    html.H1(
        "Tips Dataset Dashboard",
        style={'textAlign': 'center'}
    ),

    html.Label("Select Variable:"),

    dcc.Dropdown(
        id='variable_dropdown',
        options=[
            {'label': 'Total Bill', 'value': 'total_bill'},
            {'label': 'Tip', 'value': 'tip'},
            {'label': 'Size', 'value': 'size'}
        ],
        value='total_bill'
    ),

    dcc.Graph(
        id='histogram'
    )
])

# ------------------------------------------
# CALLBACK
# ------------------------------------------

@app.callback(
    Output('histogram', 'figure'),
    Input('variable_dropdown', 'value')
)

def update_graph(selected_variable):

    fig = px.histogram(
        df,
        x=selected_variable,
        title="Distribution of " + selected_variable
    )

    return fig


# ------------------------------------------
# RUN DASHBOARD
# ------------------------------------------

if __name__ == '__main__':
    app.run(debug=True)
```
